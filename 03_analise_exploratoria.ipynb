##### parte SQL (EDA & KPIs)

```python
conn = sqlite3.connect(':memory:')
Customer.to_sql("Customer", conn, index=False, if_exists="replace")
print("Conexão criada e tabela presente no banco")

def sql (query):
    return pd.read_sql(query,conn)
```

```python
sql("""

SELECT * 
FROM Customer

LIMIT 10

""")
```

#### Estastisticas 

###### Total vendido

```python
sql("""

SELECT 
    SUM (Total) as Total_venda
FROM fact_order

""")
```

#### Ticket médio

```python
sql("""

SELECT
    AVG(Total) as Ticket_media
FROM fact_order


""")
```

#### Maior venda

```python
sql("""
SELECT 
    MAX(Total) as Maior_venda
FROM fact_order

""")
```

#### Menor venda

```python
sql("""
SELECT 
    MIN(Total) as Menor_venda
FROM fact_order
""")
```
