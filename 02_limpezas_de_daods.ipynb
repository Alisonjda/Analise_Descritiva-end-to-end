### Limpezas de Dados

```python
import pandas as pd 
import numpy as np
import os 
import sqlite3
```

```python
Customer = pd.read_csv("archive (2)/DIM_Customer.csv", sep=",")
Delivery = pd.read_csv("archive (2)/DIM_Delivery.csv", sep=",")
Products = pd.read_csv("archive (2)/DIM_Products.csv", sep=",")
Shopping = pd.read_csv("archive (2)/DIM_Shopping.csv", sep=",")
Fact_order = pd.read_csv("archive (2)/FACT_Orders.csv", sep=",")
```

### Verificar valores nulos

##### Cópia dos dados

```python
customer_clean = Customer.copy()
delivery_clean = Delivery.copy()
products_clean = Products.copy()
shopping_clean = Shopping.copy()
orders_clean =   Fact_order.copy()
```

```python
datasets = {
    "Customer": customer_clean,
    "Delivery": delivery_clean,
    "Products": products_clean,
    "Shopping": shopping_clean,
    "Orders": orders_clean
}

for nome, df in datasets.items():
    print(f"\n===== {nome} =====")
    print(df.isnull().sum())
```

### Verificar duplicatas

```python
for nome, df in datasets.items():
    print(f"{nome}: {df.duplicated().sum()} duplicados")
```
